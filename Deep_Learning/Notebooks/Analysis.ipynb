```python
import plotly.graph_objects as go

data = [
    ["Structured Neural Network", 
     "Deep Learning (DL)", 
     "PyTorch bilan yaratilgan feedforward neural network", 
     "Input → 32 → ReLU → Dropout → 16 → ReLU → Output", 
     "BCEWithLogitsLoss, Optimizer: Adam, 50 epoch, batch size = 32"],
    
    ["RandomForestClassifier", 
     "Supervised ML (SML)", 
     "Ensemble Tree model (bagging)", 
     "Binary classification uchun", 
     "Hyperparameters: tanlangan, Metriclar: Accuracy, Precision, Recall, F1, ROC-AUC"]
]


headers = ["Algorithm", "Model Type", "Description", "Architecture / Notes", "Training / Hyperparameters"]


fig = go.Figure(data=[go.Table(
    header=dict(
        values=headers,
        fill_color='darkgreen',  
        font=dict(color='white', size=14),
        align='left'
    ),
    cells=dict(
        values=list(zip(*data)),  
        fill_color='lightcyan',    
        font=dict(color='black', size=12),
        align='left'
    )
)])

fig.update_layout(
    title="DL vs SML Model Explanations",
    title_font_size=18,
    height=350,
    width=1200,
    paper_bgcolor='#b3e0ff',  
    plot_bgcolor='#b3e0ff'    
)

fig.show()
```

```python
import plotly.express as px
import pandas as pd


data = {
    "Algorithm": ["Structured NN", "Random Forest"],
    "Accuracy": [0.8382, 0.8795],
    "Precision": [0.8068, 0.8672],
    "Recall": [0.7485, 0.8021],
    "F1 Score": [0.7766, 0.8334],
    "ROC-AUC": [0.0, 0.9513]  
}

df = pd.DataFrame(data)


df_long = df.melt(id_vars="Algorithm", var_name="Metric", value_name="Value")


def color_metric(row):
    metric = row["Metric"]
    max_val = df_long[df_long["Metric"]==metric]["Value"].max()
    min_val = df_long[df_long["Metric"]==metric]["Value"].min()
    if row["Value"] == max_val:
        return "limegreen"
    elif row["Value"] == min_val:
        return "red"
    else:
        return "steelblue"

df_long["Color"] = df_long.apply(color_metric, axis=1)


fig = px.bar(
    df_long,
    x="Metric",
    y="Value",
    color="Color",
    text="Value",
    barmode="group",
    title="Hotel Booking Cancellation: Model Comparison",
    height=500,
    width=900
)


fig.update_layout(
    plot_bgcolor='#b3ffb3',     
    paper_bgcolor='#b3ffb3',    
    font=dict(color="black"),    
    showlegend=False,
    xaxis=dict(
        title="Metrics",
        showgrid=True,
        gridcolor='rgba(0,0,0,0.1)',  
        zeroline=False,
        tickangle=-45,
        tickfont=dict(color="black")
    ),
    yaxis=dict(
        title="Value",
        showgrid=True,
        gridcolor='rgba(0,0,0,0.1)',
        zeroline=False,
        range=[0,1.05],
        tickfont=dict(color="black")
    ),
)


fig.update_traces(
    texttemplate='%{text:.4f}',
    textposition='outside',
    textfont=dict(color="black")
)

fig.show()
```
