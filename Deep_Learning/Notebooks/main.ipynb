## Model Arxitekturasi

**Structured Neural Network** (PyTorch): 

- Input Layer: 34 features
- Hidden Layer 1: 32 neyron, ReLU + Dropout(0.2)
- Hidden Layer 2: 16 neyron, ReLU
- Output Layer: 1 neyron (logit, BCEWithLogitsLoss uchun)

## Parametrlar 
- **Activation:** ReLU (hidden layerlarda)  
- **Dropout:** 0.2 (overfittingni kamaytirish uchun)  
- **Loss Function:** `BCEWithLogitsLoss` (binary classification)  
- **Optimizer:** Adam, learning rate = 0.001  


## Trening

- **Epochlar soni:** 50  
- **Batch Size:** 32  
- **Train/Validation Split:** 90% / 10%  
- **DataLoader:** Batches bilan samarali trening uchun
