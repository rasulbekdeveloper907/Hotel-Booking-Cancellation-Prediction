```python
# Notebookda ishlatish misoli
from src.data_loader import DataLoaderClass
from src.dataset import PyTorchDataset
from src.model import StructuredNN
from src.train import Trainer
from src.evaluate import Evaluator
import torch
```

```python

# 1️⃣ Ma'lumotlarni yuklash
data_loader = DataLoaderClass(
    X_train_path="../data/X_train_selected_smote.csv",
    y_train_path="../data/y_train_smote.csv",
    X_test_path="../data/X_test_selected.csv",
    y_test_path="../data/y_test.csv"
)
X_train, X_val, y_train, y_val, X_test, y_test = data_loader.load_data()
```

```python
# 2️⃣ PyTorch Dataset va DataLoader
dataset = PyTorchDataset(X_train, y_train, X_val, y_val, X_test, y_test)
train_loader, val_loader, test_loader = dataset.get_loaders()
```

```python
# 3️⃣ Model yaratish
input_dim = X_train.shape[1]
model = StructuredNN(input_dim)
```

```python
# 4️⃣ Training
trainer = Trainer(model, train_loader, val_loader, lr=0.001, num_epochs=50)
trainer.train()
trainer.plot_losses()
```

```python
# 5️⃣ Evaluation
evaluator = Evaluator(model, dataset.X_test_t, dataset.y_test_t)
evaluator.evaluate()
```
